```python
import seaborn as sns
%matplotlib inline
```

```python
flights = sns.load_dataset('flights')
```

```python
flights.head()
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
# Heatmap

tips.corr(numeric_only=True)
```

```python
sns.heatmap(tips.corr(numeric_only=True))
```

```python
sns.heatmap(tips.corr(numeric_only=True), cmap='coolwarm', annot=True)
```

```python
flights.corr(numeric_only=True)
```

```python
sns.heatmap(flights.corr(numeric_only=True))
```

```python
sns.heatmap(flights.corr(numeric_only=True), annot=True, cmap='coolwarm')
```

```python
flights.pivot_table(values='passengers', index='month', columns='year')
```

```python
pvflights = flights.pivot_table(values='passengers', index='month', columns='year')
sns.heatmap(pvflights)
```

```python
sns.heatmap(pvflights, cmap='magma', linecolor='white', lw=2)
```

```python
sns.clustermap(pvflights)
```

```python
sns.clustermap(pvflights, cmap='coolwarm', standard_scale=1, lw=2)
```
